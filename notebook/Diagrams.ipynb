```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, FancyArrowPatch
import numpy as np

# Set style
plt.style.use('default')
plt.rcParams['font.family'] = 'sans-serif'

fig, ax = plt.subplots(figsize=(6, 1.8))
ax.set_xlim(-0.5, 5.5)
ax.set_ylim(0, 0.6)
ax.axis('off')

# Steps with numbers
steps_data = [
    (0, "1", "#3498db", "Send"),
    (1, "2", "#e74c3c", "Detect"),
    (2, "3", "#f39c12", "Analyze"),
    (3, "4", "#2ecc71", "Act"),
    (4, "5", "#9b59b6", "Alert"),
]

y_main = 0.3
circle_radius = 0.12

# Draw circles and connections
for i, (x, num, color, label) in enumerate(steps_data):
    # Circle
    circle = Circle((x, y_main), circle_radius, facecolor=color, 
                    edgecolor='white', linewidth=1.5, alpha=0.9)
    ax.add_patch(circle)
    
    # Number inside circle
    ax.text(x, y_main, num, ha='center', va='center',
            fontsize=8, fontweight='bold', color='white')
    
    # Label below circle
    ax.text(x, y_main - circle_radius - 0.05, label, ha='center', va='center',
            fontsize=6, color='#2c3e50', fontweight='bold')
    
    # Arrow between circles
    if i < len(steps_data) - 1:
        arrow = FancyArrowPatch(
            (x + circle_radius, y_main),
            (x + 1 - circle_radius, y_main),
            arrowstyle='->,head_width=0.06,head_length=0.06',
            color='#95a5a6',
            linewidth=0.8
        )
        ax.add_patch(arrow)

# Title
ax.text(2.5, 0.55, 'Incident Response Pipeline', 
        ha='center', fontsize=8, fontweight='bold', color='#2c3e50')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.patches import FancyBboxPatch, Circle
import numpy as np

plt.figure(figsize=(14, 8))
ax = plt.gca()

# Colors
colors = {
    'threats': '#ff6b6b',
    'soc': '#4ecdc4',
    'response': '#45b7d1',
    'background': '#f8f9fa'
}

# Background
ax.set_facecolor(colors['background'])
plt.gcf().patch.set_facecolor(colors['background'])

# Title
plt.suptitle('Cybersecurity Threats vs SOC Workflow', fontsize=16, fontweight='bold', y=0.95)

# Left side - Threats
threats = [
    'Malware Attacks',
    'Phishing Attempts',
    'DDoS Attacks',
    'Insider Threats',
    'Zero-Day Exploits'
]

# Right side - SOC Workflow
workflow = [
    'Threat Detection',
    'Incident Analysis',
    'Impact Assessment',
    'Response Planning',
    'Remediation'
]

# Draw threat icons
threat_x = 0.2
for i, threat in enumerate(threats):
    y_pos = 0.8 - i * 0.15
    # Threat circle
    circle = Circle((threat_x, y_pos), 0.04, fill=True, color=colors['threats'], alpha=0.8)
    ax.add_patch(circle)
    plt.text(threat_x + 0.06, y_pos, threat, fontsize=10, va='center', 
             bbox=dict(boxstyle="round,pad=0.3", facecolor=colors['threats'], alpha=0.2))

# Draw SOC workflow
soc_x = 0.8
for i, step in enumerate(workflow):
    y_pos = 0.8 - i * 0.15
    # SOC rectangle
    rect = FancyBboxPatch((soc_x - 0.1, y_pos - 0.04), 0.2, 0.08, 
                         boxstyle="round,pad=0.02", 
                         facecolor=colors['soc'], alpha=0.8)
    ax.add_patch(rect)
    plt.text(soc_x, y_pos, step, fontsize=10, va='center', ha='center', 
             bbox=dict(boxstyle="round,pad=0.3", facecolor=colors['soc'], alpha=0.2))

# Arrows showing attack flow
for i in range(len(threats)):
    y_pos = 0.8 - i * 0.15
    plt.arrow(0.3, y_pos, 0.45, 0, head_width=0.02, head_length=0.02, 
              fc=colors['threats'], ec=colors['threats'], linewidth=2)

# Response feedback arrow
plt.arrow(0.85, 0.1, -0.45, 0.2, head_width=0.02, head_length=0.02,
          fc=colors['response'], ec=colors['response'], linewidth=2, linestyle='--')

plt.text(0.5, 0.25, 'Response Feedback', fontsize=9, ha='center',
         bbox=dict(boxstyle="round,pad=0.3", facecolor=colors['response'], alpha=0.3))

# Section labels
plt.text(0.2, 0.9, 'CYBERSECURITY\nTHREATS', fontsize=12, fontweight='bold', 
         ha='center', va='center', 
         bbox=dict(boxstyle="round,pad=0.5", facecolor=colors['threats'], alpha=0.3))

plt.text(0.8, 0.9, 'SOC WORKFLOW', fontsize=12, fontweight='bold', 
         ha='center', va='center',
         bbox=dict(boxstyle="round,pad=0.5", facecolor=colors['soc'], alpha=0.3))

plt.xlim(0, 1)
plt.ylim(0, 1)
plt.axis('off')

plt.tight_layout()
plt.show()
```
